```python

from sklearn.datasets import load_iris
import pandas as pd
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of Iris Dataset:")
print(df.head())

print("\n===== Measures of Central Tendency =====")

print("\nMean:")
print(df.mean())

print("\nMedian:")
print(df.median())


print("\nMode:")
print(df.mode().iloc[0])


print("\n===== Measures of Dispersion =====")

# Range
print("\nRange:")
print(df.max() - df.min())

# Variance
print("\nVariance:")
print(df.var())

# Standard Deviation
print("\nStandard Deviation:")
print(df.std())
```
